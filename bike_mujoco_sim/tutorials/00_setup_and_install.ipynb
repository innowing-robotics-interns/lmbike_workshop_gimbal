# Setup 0. Install Python and the packages (do this once)

Before the lessons can run, your computer needs three things working together:

1. **Python** itself.
2. A handful of extra **packages** for Python: MuJoCo for the physics, matplotlib for the graphs, and a few helpers.
3. An editor, such as VS Code, that knows **which** Python to use for the notebooks.

This notebook walks you through getting all three set up. You only need to do it once on each computer.

Open this notebook before Lesson 1 if any of these happen to you:

- VS Code or Jupyter says it cannot find Python.
- A cell stops with a message like `No module named 'matplotlib'`, `No module named 'mujoco'`, or `No module named 'yaml'`. This means Python itself is working, but one of the packages has not been installed yet.
- `pip install …` answers **No matching distribution**. This usually means your Python version is too new for that package.
- You have several Python versions on the computer, and you are not sure which one VS Code is using.

The steps first show the normal way to install everything by downloading it from the internet, because that is the method you will use on every future project. The last few cells contain a **failsafe script**. It does the same work automatically, for when the manual steps still leave something missing.

## A few words you will see

- **Python interpreter**: the program that actually reads and runs Python code. "Installing Python" means installing an interpreter. A computer can have several interpreters, each in a different folder.
- **Package**: an add-on that gives Python new abilities. `mujoco` is a package, and so is `matplotlib`. You install packages with a tool called `pip`.
- **PATH**: the list of folders your terminal searches when you type a command such as `python`. If the folder that holds Python is not on this list, the terminal says it cannot find Python, even though Python is installed.
- **Kernel**: the Python interpreter that a Jupyter notebook is connected to. When you run a cell, the kernel is what runs it. VS Code shows the kernel's name in the top right corner of the notebook.

## Why one test laptop needed so many retries

During a trial run on a Windows laptop, it took many attempts to get Lesson 1 working. Looking back, every problem came from one of these causes:

1. Three different Pythons were installed at the same time: one from the Microsoft Store, Python 3.14, and Python 3.11. So it was never clear which one was being used.
2. VS Code was pointed at a Python inside the Downloads or AppData folder, and that folder was not on PATH.
3. Packages were installed into one Python while the notebook was running a different one. To that notebook, the packages looked missing.
4. Someone typed `pip install yaml`. The code says `import yaml`, but the package that provides it is named **PyYAML**, so that command fails.
5. Python 3.14 is so new that matplotlib and MuJoCo have not yet published ready-made installs for it.

All five have the same cure, which is one simple rule: **use one Python 3.10 install, one virtual environment for this project, and select that same environment as the notebook's kernel.** The steps below set up exactly that.

## Step B. Create a virtual environment for this project

A **virtual environment** is a private copy of Python that belongs to this one project. It lives in a folder named `.venv` inside `bike_mujoco_sim`. While the environment is switched on, every package you install goes into that folder and nowhere else. This has two benefits. Packages for this course cannot clash with packages from another course. And if something ever breaks, you can delete the `.venv` folder and create a fresh one in a minute.

Open a terminal and move into the `bike_mujoco_sim` folder, which is the folder that contains `run_sim.py`. Then type the commands for your system. Here is what they do, in order:

1. Go into the project folder.
2. Create the environment.
3. Switch it on (this is called *activating* it).
4. Print the Python version, so you can check it is 3.10.
5. Update `pip`, the package installer.

### Windows PowerShell

```powershell
cd path\to\bike_mujoco_sim
py -3.10 -m venv .venv
.\.venv\Scripts\Activate.ps1
python --version
python -m pip install --upgrade pip
```

If Windows refuses to run `Activate.ps1` with a message about running scripts being disabled, run this command once, then try the activate command again:

```powershell
Set-ExecutionPolicy -Scope CurrentUser RemoteSigned
```

### macOS / Linux

```bash
cd path/to/bike_mujoco_sim
python3.10 -m venv .venv
source .venv/bin/activate
python --version
python -m pip install --upgrade pip
```

When the environment is switched on, the start of your terminal prompt shows `(.venv)`. That is your sign that any `pip install` you type now will go into this project's private Python.

## Step C. Point VS Code and Jupyter at that Python

The environment exists now, but VS Code does not use it automatically. You need to tell VS Code to run the notebooks with the Python inside `.venv`. That choice is the notebook's kernel.

1. Open the `bike_mujoco_sim` folder in VS Code with **File → Open Folder**.
2. Open this notebook, `tutorials/00_setup_and_install.ipynb`.
3. Click the kernel name in the top right corner of the notebook.
4. Choose **Select Another Kernel… → Python Environments…**
5. Pick the entry whose path contains `bike_mujoco_sim\.venv\` on Windows, or `bike_mujoco_sim/.venv/` on macOS and Linux.

Here is how to tell whether you picked the right one. The right entry has both of these:

- a path that ends with `.venv\Scripts\python.exe` (Windows) or `.venv/bin/python` (macOS and Linux)
- version 3.10.x or 3.11.x

Choose a different entry if you see any of the following, because each one points to some other Python on the computer, not the project's own:

- a path inside `Downloads`
- a path inside `WindowsApps`, which is the Microsoft Store Python
- version 3.14.x

If the `.venv` entry is not in the list, you can type its location by hand. Open the Command Palette (`Ctrl+Shift+P` on Windows and Linux, `Cmd+Shift+P` on macOS), choose **Python: Select Interpreter**, then **Enter interpreter path…**, and browse to the `.venv` Python described above.

## Step D. Check what this notebook is actually using

The next cell asks the kernel three questions: which Python version it is, where that Python lives on the disk, and whether it can load each package the lessons need. Run it and read what it prints.

- If the version is **3.14**, or the path is outside `.venv`, go back to Step C and select the `.venv` kernel.
- If the version and path look right but some lines start with `MISS`, those packages are not installed yet. Continue to Step E to install them by hand, or to Step F for the automatic script.
- If every package line starts with `OK`, everything is ready. Go straight to Step G.

````python
import os
import sys

print("Python version :", sys.version)
print("Executable     :", sys.executable)
print("Prefix         :", sys.prefix)

ver = sys.version_info
if ver >= (3, 14):
    print("PROBLEM: Python 3.14+ is too new for this class. Install 3.10 and select that interpreter.")
elif ver < (3, 10):
    print("PROBLEM: need Python 3.10 or 3.11.")
else:
    print("Version looks OK.")

in_venv = (
    hasattr(sys, "real_prefix")
    or (hasattr(sys, "base_prefix") and sys.base_prefix != sys.prefix)
    or os.path.basename(os.path.dirname(sys.prefix)) == ".venv"
    or os.path.basename(sys.prefix) == ".venv"
)
print("Looks like a virtual env:", bool(in_venv))

needed = [
    ("numpy", "numpy"),
    ("scipy", "scipy"),
    ("matplotlib", "matplotlib"),
    ("yaml", "PyYAML"),
    ("mujoco", "mujoco"),
    ("ipykernel", "ipykernel"),
]
missing = []
for module_name, pip_name in needed:
    try:
        mod = __import__(module_name)
        version = getattr(mod, "__version__", "?")
        print(f"OK   {module_name:12} {version}")
    except Exception as exc:
        missing.append(pip_name)
        print(f"MISS {module_name:12} -> install package named {pip_name!r} ({exc.__class__.__name__})")

if missing:
    print("\nMissing packages:", ", ".join(missing))
    print("Remember: import yaml  comes from  pip package PyYAML  (not 'yaml').")
else:
    print("\nAll required packages import cleanly. You can open Lesson 1.")
````

## Step E. Install the packages the traditional way (in a terminal)

The file `requirements.txt` in `bike_mujoco_sim` lists every package the lessons need. A single `pip` command can read that list and install everything on it.

In the terminal where `(.venv)` is showing, type the three commands below. Here is what each one does:

1. Update the installer tools themselves.
2. Install everything listed in `requirements.txt`.
3. Register this environment with Jupyter under the friendly name "Bike MuJoCo sim", so it is easy to find in the kernel menu.

### Windows PowerShell

```powershell

python -m pip install -r requirements.txt
python -m ipykernel install --user --name bike-mujoco-sim --display-name "Bike MuJoCo sim"
```

### macOS / Linux

```bash
python -m pip install --upgrade pip setuptools wheel
python -m pip install -r requirements.txt
python -m ipykernel install --user --name bike-mujoco-sim --display-name "Bike MuJoCo sim"
```

When the commands finish, scroll up and run the check cell in Step D again. Every package line should now start with `OK`.

### If a package name confuses you

The name you `import` in Python is sometimes different from the name you install with `pip`. These are the cases that tripped people up before:

| You tried | Use this instead |
| --- | --- |
| `pip install yaml` | `pip install PyYAML` |
| `pip install mujoco-py` | `pip install mujoco` |
| Store Python, or Python 3.14 | Install Python 3.10 from python.org, then create `.venv` again |

## Step F. Failsafe: run the automatic installer from this notebook

Use this step if Steps A to E still leave some packages missing **and** the check cell in Step D already shows Python 3.10 or 3.11 inside `.venv`. The script installs packages into whichever Python runs it, so it can only help once the right kernel is selected.

The next cell runs `tutorials/setup_env.py` using this notebook's own Python. The script does the same three jobs as Step E for you: it updates `pip`, installs everything in `requirements.txt`, and registers the Jupyter kernel. Run the cell and wait for the line `exit code: 0`, which means the script finished without an error.

````python
import os
import subprocess
import sys

def find_project_root(start=None):
    current = os.path.abspath(start or os.getcwd())
    while True:
        if os.path.isfile(os.path.join(current, "run_sim.py")) and os.path.isfile(
            os.path.join(current, "requirements.txt")
        ):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            break
        current = parent
    raise RuntimeError("Could not find bike_mujoco_sim. Open the folder that contains run_sim.py.")

root = find_project_root()
script = os.path.join(root, "tutorials", "setup_env.py")
print("Using:", sys.executable)
print("Running:", script)
result = subprocess.run([sys.executable, script], cwd=root)
print("exit code:", result.returncode)
if result.returncode != 0:
    raise RuntimeError(
        "Automatic setup failed. Go back to Step A: install Python 3.10 from python.org, "
        "create .venv, select that interpreter, then run this cell again."
    )
````

### Still stuck?

These clean-up steps fix almost every remaining case:

- Keep a single Python 3.10 and uninstall the extra ones, such as Microsoft Store copies or 3.14. With fewer Pythons on the computer, there are fewer wrong ones to pick by accident.
- Delete the old `.venv` folder and create a fresh one with `py -3.10 -m venv .venv` (Windows) or `python3.10 -m venv .venv` (macOS and Linux). Then repeat Step E.
- In VS Code's settings, search for `python.defaultInterpreterPath`. If it points into Downloads, clear it.
- Restart VS Code after any change to PATH, so that it picks up the new list of folders.

If you ticked **Add python.exe to PATH** in the installer and you use `.venv`, you will not need to edit any environment variables by hand.
