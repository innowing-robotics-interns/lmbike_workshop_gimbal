## Glossary

Read this section once. Later steps use these words without teaching them again.

### F = ma

A push is a force, F. The object's mass is m. How quickly it speeds up is a, the acceleration. A harder push on the same object makes it speed up faster. Picture a shopping cart: the same cart accelerates more when you push harder.

### Physics engine

MuJoCo is the physics engine for this lesson. It keeps doing F = ma for the bicycle, the ground, and the wheels, many times a second. It is like a video game that obeys pushes and gravity. It does not decide how to steer. The 3D window is the MuJoCo window.

### Free-body diagram

A free-body diagram is a sketch of one object and the arrows of the forces on it. Here the object is the bicycle. Gravity points down. The ground pushes up. The motors twist.

### F = −kx

A spring pulls back harder when it is stretched more. x is the stretch you measure. k is how stiff the spring is. The minus sign means the pull is back toward the start. k is a gain: the same stretch, a bigger k, a harder pull.

### Gain

A gain is that number k. It scales a reaction. The measurement stays the same, and a bigger gain makes a stronger command. The gain this lesson changes is `pid_kd_roll`. It is the same kind of multiplier, but it multiplies how fast the lean is changing, not how far the bicycle has already leaned.

### PID

PID is three reactions added together.

- **Proportional (P).** The command grows with the lean right now. A 10° lean gets a stronger command than a 2° lean. The setting is `pid_kp_roll`.
- **Integral (I).** The command grows if that lean keeps lasting. A small lean that stays for a long time slowly adds extra command. The setting is `pid_ki_roll`.
- **Derivative (D).** The command grows with how fast the lean is changing. Two bicycles at the same 5° lean get different commands if one is tipping faster. The setting is `pid_kd_roll`. This lesson changes that number.

`pid_` is the family name: proportional, integral, derivative. From here on, the lesson says P, I, and D.

| Term | Meaning | Example |
| --- | --- | --- |
| Simulation | The physics engine playing the free-body diagram forward in tiny time steps, fast enough to look like a ride. | Many F = ma steps, played as motion. |
| Model | One bicycle in that simulation: its weight, its wheels, and how the parts are joined. This lesson uses `new_bike`. | The object drawn in the free-body diagram. |
| Roll | How far the bicycle leans sideways, in degrees. 0° means upright. | Stand a book on its edge. The angle as it starts to fall is the roll. |
| Controller | The balancing loop that looks at the lean and tells the steering motor what to do. The physics engine does not do this job. | Balancing a stick on your hand. Your eyes see the tilt. Your hand moves to catch it. |
| Torque | A twist from a motor. Steering torque is what turns the handlebars. | Your hand twisting a doorknob. |
| Configuration | The list of numbers for this ride. The cell in Step 3 prints that list. | The settings on a phone. The phone stays the same. The settings change what it does. |
| Reference | The target. The real bicycle can miss it. In the printout, target names end in `_ref`. | The temperature you set on an oven. The food is not that hot yet. |
| Speed schedule | How the bicycle's velocity changes with time. The rear wheel is what gives the bicycle that forward velocity. | The bicycle speeds up, then keeps one velocity. That velocity does not stop the bicycle from leaning. |

## Step 2. Look at the bicycle's velocity

A speed schedule is how the bicycle's velocity is chosen ahead of time. The bicycle starts at 0 m/s, so it has no forward velocity yet. Over `acceleration_time` seconds its velocity rises to `target_speed`. After that, the bicycle keeps that same velocity. That steady velocity is the cruise speed: the velocity after the speed-up is finished. Read the number the cell prints next to `target_speed`.

The rear wheel is what gives the bicycle this forward velocity. The velocity says nothing about the bicycle leaning sideways. It is like cruise control in a car. Cruise control holds a velocity. It does not keep the bicycle from leaning.

The cell prints the requested velocity before you watch the bicycle. A reference is the target. The bicycle can miss it.

- `speed_ref` is the speed we want, in meters per second. It should rise, then stay at the cruise speed.
- `roll_ref` is the lean we want, in degrees. It prints as 0, which means upright.
- `steer_ref` is the handlebar angle we want. It prints as 0, which means straight ahead.

The settings are read from a text file whose name ends in `.yaml`. In the cell, `speed_goal=None` means "give the bicycle this velocity." It does not mean the bicycle is asked to have no velocity.

Run the next cell. You are done when `speed_ref` rises and then sits at the printed `target_speed`, and both `roll_ref` and `steer_ref` print as 0. Those zeros are the request. They are not a measurement of the bicycle.


```python
new_bike_yaml = os.path.join(project_root, "configs", "new_bike.yaml")
cfg = load_config(new_bike_yaml)
print("target_speed", cfg.target_speed, "m/s")
print("acceleration_time", cfg.acceleration_time, "s")
for t in (0.0, 0.5, 1.0, cfg.acceleration_time, cfg.acceleration_time + 1.0):
    speed_ref, roll_ref, steer_ref = apply_speed_refs(
        mode="speed-schedule",
        time_s=t,
        speed_goal=None,
        steer_ref=0.0,
        cfg=cfg,
    )
    print(f"t={t:4.2f}s  speed_ref={speed_ref:.3f}  roll_ref={roll_ref}  steer_ref={steer_ref}")
```

## Step 3. Watch the bicycle

You already met the gain k in F = −kx. Same stretch, bigger k, harder pull. `pid_kd_roll` is that kind of number, but the measurement is how fast the lean is changing.

Picture two bicycles, both leaned over by 5°. One is tipping slowly and one is tipping quickly. Proportional looks at the 5° and treats them the same. Derivative looks at the tip speed and treats them differently. `pid_kd_roll` is the derivative gain, k for the fast tip. This ride sets it to 1, which makes that answer small.

The controller is the balancing loop: the PID that chooses the steering command. The physics engine only moves the bicycle.

The next cell prints the configuration. That printed list is the settings for this ride. You do not need to open another file to see them.

| Name | What it is |
| --- | --- |
| `target_speed` | The cruise speed, in m/s, after the short speed-up. |
| `acceleration_time` | How many seconds that speed-up lasts. |
| `duration` | How many seconds this ride lasts. |
| `pid_kp_roll` | Proportional gain. How hard to react to the lean right now. |
| `pid_ki_roll` | Integral gain. How hard to react to a lean that is still there after the first moment. |
| `pid_kd_roll` | Derivative gain. How hard to react when the lean is changing fast. This ride uses 1. |
| `pid_kp_steer` | How hard to bring the handlebars back to straight. |
| `steer_kp`, `steer_ki` | A second, smaller control loop inside the steering motor. Leave them as printed. |

The bicycle is sent toward a steady speed. The speed-up takes `acceleration_time`. After that, the requested speed stays the same. The balancing loop stays on for the whole ride. This ride does not turn that loop off.

You do not need to edit the cell.

```python
import os
import sys

def find_project_root(start=None):
    current = os.path.abspath(start or os.getcwd())
    while True:
        if os.path.isfile(os.path.join(current, "run_sim.py")) and os.path.isfile(
            os.path.join(current, "sim", "__init__.py")
        ):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            break
        current = parent
    raise RuntimeError(
        "Could not find bike_mujoco_sim. Start Jupyter from the folder that "
        "contains run_sim.py, or from its tutorials folder."
    )

project_root = find_project_root()
if project_root not in sys.path:
    sys.path.insert(0, project_root)

if "mujoco" in sys.modules and os.environ.get("MUJOCO_GL") not in (None, "", "glfw"):
    raise RuntimeError(
        "MuJoCo was already imported with MUJOCO_GL=%r. Restart the kernel, "
        "then run this cell again." % os.environ.get("MUJOCO_GL")
    )
os.environ["MUJOCO_GL"] = "glfw"

from dataclasses import replace
from sim.config import load_config
from sim.loop import make_sim
from sim.viewer import run_viewer

print("project_root:", project_root)
print("MUJOCO_GL:", os.environ.get("MUJOCO_GL"))


new_bike_yaml = os.path.join(project_root, "configs", "new_bike.yaml")
cfg = load_config(new_bike_yaml)
cfg = replace(cfg, pid_kd_roll=1.0, duration=4.0)
sim = make_sim(model_spec="new_bike", mode="speed-schedule", cfg=cfg)
print("configuration for this ride")
print(f"  file                 {os.path.relpath(new_bike_yaml, project_root)}")
print(f"  target_speed         {cfg.target_speed} m/s")
print(f"  acceleration_time    {cfg.acceleration_time} s")
print(f"  duration             {cfg.duration} s")
print(f"  pid_kp_roll          {cfg.pid_kp_roll}")
print(f"  pid_ki_roll          {cfg.pid_ki_roll}")
print(f"  pid_kd_roll          {cfg.pid_kd_roll}  (set to 1 for this ride)")
print(f"  pid_kp_steer         {cfg.pid_kp_steer}")
print(f"  steer_kp             {cfg.steer_kp}")
print(f"  steer_ki             {cfg.steer_ki}")
print("Close the MuJoCo window to finish this cell.")
run_viewer(sim)
print("Window closed.")
```

## Step 4. Draw the lean

The next cell repeats the ride you just watched. It does not open a window. It records the lean and draws it. Time runs to the right. Lean runs up and down. The flat line at 0° is upright.

Run the cell. When the graph appears, read the next gray cell. That cell is about the line you are looking at.


```python
new_bike_yaml = os.path.join(project_root, "configs", "new_bike.yaml")
cfg = load_config(new_bike_yaml)
cfg = replace(cfg, pid_kd_roll=1.0, duration=4.0)
sim = make_sim(model_spec="new_bike", mode="speed-schedule", cfg=cfg)
print("same ride as the window, pid_kd_roll =", cfg.pid_kd_roll)
history = run_headless(sim, duration=4.0)

time_s = np.asarray(history["time"], dtype=float)
roll_deg = np.degrees(np.asarray(history["roll"], dtype=float))
max_abs = float(np.max(np.abs(roll_deg)))
tip = np.where(np.abs(roll_deg) > 45.0)[0]
tip_t = float(time_s[tip[0]]) if len(tip) else None

fig, ax = plt.subplots(figsize=(8, 3.5), constrained_layout=True)
ax.plot(time_s, roll_deg)
ax.axhline(45.0, color="0.5", linestyle="--", linewidth=0.8)
ax.axhline(-45.0, color="0.5", linestyle="--", linewidth=0.8)
if tip_t is not None:
    ax.axvline(tip_t, color="C3", linestyle=":", label="lean reaches 45°")
    ax.legend()
ax.set_xlabel("time (s)")
ax.set_ylabel("roll (deg)")
ax.set_title("pid_kd_roll = 1")
plt.show()

if max_abs < 45.0:
    raise RuntimeError("expected this ride to fall")
```

## The ride, drawn as a line

The curve above is the ride you watched, drawn as lean versus time. The left end is the start. Moving right is later. Up and down is the lean, the same roll angle as the book tipping on its edge. The title is the derivative gain used for this ride, `pid_kd_roll = 1`.

The flat line at 0° is upright. A curve that stays within a few degrees of that line is a bicycle that stayed nearly upright. The plus or minus sign says which side. The size of the number says how far.

This lesson uses one mark for a fall: the dashed lines at +45° and −45°. They are not a path the bicycle was asked to follow. In Step 2 the request was `roll_ref = 0`, the upright line, not the dashed lines. If the curve crosses a dashed line, the lean counts as a fall. The dotted vertical line, if you see one, marks the first time that happened. Read that time along the bottom of the graph.

A lean past about 90° means the bicycle has gone past sideways and is on its way to lying on the ground. A number like 117° is that kind of fall.

The window showed the motion. This curve is the lean from that same motion.

## Discussion

### Why did the bicycle tip if a balancing loop was already running?

The balancing loop was on for the whole ride. This ride did not turn it off. `pid_kd_roll` was 1. That is a small k on tip speed: a fast tip produces only a weak steering command. Look back at the curve and at the dashed lines at 45°.

### Why did the bicycle's velocity not stop the bicycle from leaning and falling over?

The rear wheel gives the bicycle a forward velocity and then holds that velocity. A forward velocity does not push the bicycle sideways, so it does not stop the bicycle from leaning. `roll_ref = 0` was the lean we wanted for the bicycle. Wanting the bicycle upright at 0° does not by itself create the steering torque that would stop the bicycle from leaning.

### Does the size of that derivative gain matter?

Yes. The bicycle can keep the same velocity, and the balancing loop can stay the same, while `pid_kd_roll` changes. The questions asked you to raise that k and compare whether the bicycle keeps leaning.
