# PID and the struggle

This notebook is about **control**: extra software that looks at the bike and
tells the steering motor what to do.

You do not need a 3D window here. We will:

1. Explain PID in everyday language
2. Show the exact Python that implements it
3. Run a **weak** controller that tries, then falls
4. Point at a settings file you can edit

## Words we will use

- **Roll**: how far the bike has leaned, in degrees. 0 is upright.
- **Error**: "what we want minus what we have." We want roll = 0, so if the
  bike leans +3°, the error is −3° (we need to go the other way).
- **Gain**: a multiplier. Bigger gain means a stronger reaction to the same
  error.
- **Torque**: a twist the motor applies. Steering torque turns the handlebars.
- **Steer rate**: how fast the handlebars are turning, in radians per second.
  (About 1 rad/s is roughly 57° per second.)

## The two-loop idea (cascade)

A human balancing a broom does two things: decide which way to move the hand,
then actually move the hand. This controller is similar:

- **Outer loop (PID)** looks at lean and says "turn the bars at this speed."
- **Inner loop (PI)** looks at the actual bar speed and applies motor torque
  until that speed happens.

We will mostly talk about the outer loop. The inner loop is already set and
you should not change it in this lesson.


## P, I, and D in plain language

PID is three reactions added together.

**P — Proportional.** "The more I lean, the harder I steer." Like pushing
harder on a door the farther it is from closed. If P is too small, the bike
does not fight enough. If P is huge and there is no damping, it can overshoot
and wobble.

**D — Derivative.** "If I am falling *faster*, steer harder now." D looks at
**roll rate** (how quickly the lean is changing), not only the lean itself.
This is the term that usually **catches a fall**. On this bike, D is the
difference between tipping and staying up.

**I — Integral.** "If a small lean lasts a long time, keep adding extra
command." I is for leftover lean that P and D leave sitting there. It is
slow. We clip it so it cannot grow forever.

There is also a **steer-angle term**: "bring the handlebars back toward
straight." That keeps the bars from wandering.

```text
sensors (lean, bar angle)
        ↓
outer PID  →  desired bar speed
        ↓
inner PI   →  steering motor torque
        ↓
handlebars actually turn
```


````python
import os
import sys

def choose_mujoco_gl(platform, display, existing):
    if existing:
        return existing
    if str(platform).startswith("linux") and not display:
        return "egl"
    return None

chosen = choose_mujoco_gl(sys.platform, os.environ.get("DISPLAY"), os.environ.get("MUJOCO_GL"))
if chosen and "MUJOCO_GL" not in os.environ:
    os.environ["MUJOCO_GL"] = chosen

def find_project_root(start=None):
    current = os.path.abspath(start or os.getcwd())
    while True:
        if os.path.isfile(os.path.join(current, "run_sim.py")) and os.path.isfile(
            os.path.join(current, "sim", "__init__.py")
        ):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            break
        current = parent
    raise RuntimeError(
        "Could not find bike_mujoco_sim. Start Jupyter from the folder that "
        "contains run_sim.py, or from its tutorials folder."
    )

project_root = find_project_root()
if project_root not in sys.path:
    sys.path.insert(0, project_root)

%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from dataclasses import replace

from sim.config import load_config
from sim.loop import make_sim, run_headless, summarize
from sim.pid_controller import RollSteerPID, apply_speed_refs

np.set_printoptions(precision=4, suppress=True)
print("project_root:", project_root)
print("platform:", sys.platform, "MUJOCO_GL:", os.environ.get("MUJOCO_GL", "(unset)"))


from sim.sensors import wrap_angle
````

## Walking through one PID step

The class `RollSteerPID` lives in `sim/pid_controller.py`. Each 0.02 s
(that is 50 times per second) it does:

1. **Roll error** = wanted roll − actual roll. Wanted roll is 0.
2. **Steer error** = wanted bar angle − actual bar angle (wrapped so a full
   spin still looks like a small angle).
3. **Add to the integral**, then clip it so it cannot explode.
4. **Add the four terms** (P, I, D, steer).
5. **Limit** the result so we never ask the bars to spin faster than
   `max_steer_velocity`.

The next cell uses a fake lean of **+3°** and a fake fall rate of **+20°/s**,
then prints each term. You should see D doing a lot of the work because the
lean is growing quickly. I will be small on the first step because the
integral has barely started.

These printed units are **radians per second** of desired bar speed, not
degrees, and not motor torque yet.


````python
tutorial_yaml = os.path.join(project_root, "configs", "tutorial.yaml")
cfg = load_config(tutorial_yaml)
pid = RollSteerPID(
    kp_roll=cfg.pid_kp_roll,
    ki_roll=cfg.pid_ki_roll,
    kd_roll=cfg.pid_kd_roll,
    kp_steer=cfg.pid_kp_steer,
    max_steer_velocity=cfg.max_steer_velocity,
    integral_limit=cfg.pid_integral_limit,
)

roll = float(np.deg2rad(3.0))
roll_rate = float(np.deg2rad(20.0))
steer = 0.0
dt = cfg.control_dt
e_roll = 0.0 - roll
e_steer = 0.0 - wrap_angle(steer)
pid.integral = 0.0
pid.integral = max(-pid.integral_limit, min(pid.integral_limit, pid.integral + e_roll * dt))
p_term = pid.kp_roll * e_roll
i_term = pid.ki_roll * pid.integral
d_term = pid.kd_roll * (-roll_rate)
s_term = pid.kp_steer * e_steer
raw = p_term + i_term + d_term + s_term
cmd = float(max(-pid.max_steer_velocity, min(pid.max_steer_velocity, raw)))
print("roll = +3 deg, roll_rate = +20 deg/s")
print(f"  P term {p_term:+.3f} rad/s")
print(f"  I term {i_term:+.3f} rad/s")
print(f"  D term {d_term:+.3f} rad/s")
print(f"  steer term {s_term:+.3f} rad/s")
print(f"  clipped command {cmd:+.3f} rad/s")
print("library step()", pid.step(steer=steer, roll=roll, roll_rate=roll_rate, dt=dt))
````

## The inner loop (you will not tune this here)

The outer PID only *asks* for a bar speed. The motor still needs a twist
command. In `sim/loop.py` the inner loop is:

1. Compare desired bar speed to the measured bar speed
2. Integrate that speed error
3. Torque = `steer_kp × error + steer_ki × integral`

`steer_kp` and `steer_ki` are **motor** knobs. Changing them without care
can slam the motor into its torque limit. Leave them as they are. Your
balance knobs are the outer `pid_*` names.


## The settings file (YAML)

A **YAML file** is a plain-text list of names and numbers. Ours lives at
`configs/tutorial.yaml`. Open it in the editor next to this notebook.

The four names you may change:

| Name in the file | Meaning |
| --- | --- |
| `pid_kp_roll` | How hard to fight the current lean (P) |
| `pid_ki_roll` | How hard to fight a lean that lasts (I) |
| `pid_kd_roll` | How hard to fight a *fast* fall (D) |
| `pid_kp_steer` | How hard to straighten the bars |

The file **starts weak on purpose**. The controller will twitch, then the
bike will still fall. That is the "struggle" this lesson wants you to see.

Do not change `steer_kp` or `steer_ki` here.


````python
tutorial_yaml = os.path.join(project_root, "configs", "tutorial.yaml")
cfg = load_config(tutorial_yaml)
print("loaded", os.path.relpath(tutorial_yaml, project_root))
print(
    "outer gains: kp_roll={:.3f} ki_roll={:.3f} kd_roll={:.3f} kp_steer={:.3f}".format(
        cfg.pid_kp_roll, cfg.pid_ki_roll, cfg.pid_kd_roll, cfg.pid_kp_steer
    )
)
print("inner motor: steer_kp={:.4f} steer_ki={:.4f}".format(cfg.steer_kp, cfg.steer_ki))
````

## Run the weak controller and plot it

The next cell loads `tutorial.yaml`, runs 6 seconds of physics **without** a
window, and draws two graphs stacked:

1. **Roll versus time.** If this crosses ±45°, the bike has fallen. A red
   dotted line marks that moment.
2. **Steer-rate command versus time.** This is what the outer PID *asked*
   the bars to do. If it hits a flat ceiling, the command is saturated
   (asking for the maximum allowed bar speed).

Watch the first second closely. You should see roll grow, the command move,
then roll explode through 45°.


````python
tutorial_yaml = os.path.join(project_root, "configs", "tutorial.yaml")
cfg = load_config(tutorial_yaml)
cfg = replace(cfg, duration=6.0)
sim = make_sim(model_spec="new_bike", mode="speed-schedule", cfg=cfg)
history = run_headless(sim, duration=6.0)
summarize(history, model_path=sim.model_path, mode="struggle")

time_s = np.asarray(history["time"], dtype=float)
roll_deg = np.degrees(np.asarray(history["roll"], dtype=float))
rate_cmd = np.asarray(history["steer_rate_ref"], dtype=float)
tip = np.where(np.abs(roll_deg) > 45.0)[0]
tip_t = float(time_s[tip[0]]) if len(tip) else None

fig, axes = plt.subplots(2, 1, figsize=(8, 6), sharex=True, constrained_layout=True)
axes[0].plot(time_s, roll_deg)
axes[0].axhline(45, color="0.5", linestyle="--", linewidth=0.8)
axes[0].axhline(-45, color="0.5", linestyle="--", linewidth=0.8)
if tip_t is not None:
    axes[0].axvline(tip_t, color="C3", linestyle=":", label=f"tip ~ {tip_t:.2f}s")
    axes[0].legend()
axes[0].set_ylabel("roll (deg)")
axes[0].set_title("Weak PID — fights, then falls")
axes[1].plot(time_s, rate_cmd)
axes[1].set_ylabel("steer-rate command (rad/s)")
axes[1].set_xlabel("time (s)")
plt.show()

print(f"max |roll| = {float(np.max(np.abs(roll_deg))):.1f} deg")
if tip_t is None:
    raise RuntimeError("expected this weak gain set to tip")
# Show a few samples before the tip for the write-up.
for ti in (0.2, 0.4, 0.6, 0.8):
    i = int(np.argmin(np.abs(time_s - ti)))
    if time_s[i] <= tip_t + 0.05:
        print(f"t={time_s[i]:.2f}s  roll={roll_deg[i]:+6.1f} deg  rate_cmd={rate_cmd[i]:+.2f} rad/s")
````

## How to read what happened

Starting gains in `tutorial.yaml` are about:

- P (`pid_kp_roll`) = 5
- I (`pid_ki_roll`) = 0
- D (`pid_kd_roll`) = 6

On this bike that is **not enough D**. Roll goes from a few degrees to tens
of degrees in under a second. The steer command does move — the controller
is not asleep — and it often hits the speed limit. The lean still
accelerates. Gravity is winning the race.

This is **not** a long wobble from too much P. It is a short fight, then a
tip near 0.8 s. In words: the bike tries to steer into the lean, but it
does not brake the fall rate hard enough.

If you later make P very large without enough D, the graph can look more
like a growing oscillation (lean left, right, left, bigger each time). For
*this* file, look for a one-way dive after a brief struggle.

Those four `pid_*` numbers are what you will change to make the bike stay
up. Keep the inner motor numbers fixed.
