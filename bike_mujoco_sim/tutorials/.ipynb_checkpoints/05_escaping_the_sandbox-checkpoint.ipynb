# Leaving the notebook (running the real program)

Jupyter is a good classroom: you can read, run a little code, and see a
graph. It is a **bad** place to run a real robot or even a long-lived
simulator.

This file shows where the **same** simulator lives as ordinary project
files, how to save your numbers, and how to start it from a terminal like
an engineer would.

Run the next cell first so Python can find the project folder.


````python
import os
import sys

def choose_mujoco_gl(platform, display, existing):
    if existing:
        return existing
    if str(platform).startswith("linux") and not display:
        return "egl"
    return None

chosen = choose_mujoco_gl(sys.platform, os.environ.get("DISPLAY"), os.environ.get("MUJOCO_GL"))
if chosen and "MUJOCO_GL" not in os.environ:
    os.environ["MUJOCO_GL"] = chosen

def find_project_root(start=None):
    current = os.path.abspath(start or os.getcwd())
    while True:
        if os.path.isfile(os.path.join(current, "run_sim.py")) and os.path.isfile(
            os.path.join(current, "sim", "__init__.py")
        ):
            return current
        parent = os.path.dirname(current)
        if parent == current:
            break
        current = parent
    raise RuntimeError(
        "Could not find bike_mujoco_sim. Start Jupyter from the folder that "
        "contains run_sim.py, or from its tutorials folder."
    )

project_root = find_project_root()
if project_root not in sys.path:
    sys.path.insert(0, project_root)

%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from dataclasses import replace

from sim.config import load_config
from sim.loop import make_sim, run_headless, summarize
from sim.pid_controller import RollSteerPID, apply_speed_refs

np.set_printoptions(precision=4, suppress=True)
print("project_root:", project_root)
print("platform:", sys.platform, "MUJOCO_GL:", os.environ.get("MUJOCO_GL", "(unset)"))
````

## A map of the project (plain language)

Think of `bike_mujoco_sim` as a small workshop with labeled drawers:

| Drawer | File | What it is |
| --- | --- | --- |
| Front door | `run_sim.py` | The program you start. Like "press play." |
| Settings | `configs/tutorial.yaml` | Your numbers (P, I, D, speeds). A text file, not code. |
| Factory settings | `configs/new_bike.yaml` | A known-good copy. Do not overwrite this while learning. |
| Brain of the PID | `sim/pid_controller.py` | The math: error → desired bar speed. |
| Heartbeat | `sim/loop.py` | 50 times per second: read sensors, run PID, write motors, step physics. |
| Window | `sim/viewer.py` | The 3D view and keyboard. |
| Stick | `sim/joystick.py` | The on-screen joystick and gamepad reading. |
| Loading the bike | `sim/models.py` | Finds the 3D model file and checks motors/sensors exist. |
| The bike itself | `models/new_bike/new_bike.xml` | MuJoCo's description of masses, wheels, and joints. |

The next cell checks that each of those files exists on your computer.


````python
paths = [
    os.path.join(project_root, "run_sim.py"),
    os.path.join(project_root, "configs", "tutorial.yaml"),
    os.path.join(project_root, "configs", "new_bike.yaml"),
    os.path.join(project_root, "sim", "pid_controller.py"),
    os.path.join(project_root, "sim", "loop.py"),
    os.path.join(project_root, "sim", "viewer.py"),
    os.path.join(project_root, "sim", "joystick.py"),
    os.path.join(project_root, "sim", "models.py"),
    os.path.join(project_root, "models", "new_bike", "new_bike.xml"),
]
for path in paths:
    rel = os.path.relpath(path, project_root)
    print(("OK " if os.path.isfile(path) else "MISSING "), rel)
````

## Save your numbers on disk

Notebooks forget. If you only changed numbers in a Python variable, they
vanish when you close the kernel. Files do not.

Open `configs/tutorial.yaml` in the editor. Set:

```yaml
pid_kp_roll: 7.535559389256096
pid_ki_roll: 1.2614702013440495
pid_kd_roll: 11.260517241323017
pid_kp_steer: 1.1164231977942145
```

Save. Then run the next cell. It **reads the file from disk** and prints
what it found. If you still see 5 / 0 / 6, the file was not saved.


````python
tutorial_yaml = os.path.join(project_root, "configs", "tutorial.yaml")
cfg = load_config(tutorial_yaml)
print("File:", os.path.relpath(tutorial_yaml, project_root))
print(
    "pid_kp_roll={}\npid_ki_roll={}\npid_kd_roll={}\npid_kp_steer={}".format(
        cfg.pid_kp_roll, cfg.pid_ki_roll, cfg.pid_kd_roll, cfg.pid_kp_steer
    )
)
````

## Why engineers do not fly robots from Jupyter

**Cells run in any order.** You can skip a cell or run an old one twice.
Then a leftover variable silently changes the answer. That is called
**hidden state**. On a robot, hidden state is dangerous.

**A notebook is not a clock.** A real controller must wake up every 0.02 s,
every time, until you stop it. Jupyter waits for you to click.

**Version control is messy.** Git can track `.py` and `.yaml` cleanly. A
notebook stores huge output blobs and is hard to review.

A **modular program** is the opposite:

- `run_sim.py` is one front door
- each `.py` file has one job
- numbers live in YAML that you can diff
- the loop in `sim/loop.py` owns time

Learn in Jupyter. Ship from files and a terminal.


## Close this notebook and start the program

1. Make sure `configs/tutorial.yaml` is saved.
2. You may leave Jupyter running, or stop it with Ctrl+C in its terminal.
3. Open the `bike_mujoco_sim` folder in VS Code (or another editor), **or**
   open a new terminal and `cd` into that folder.
4. Activate the virtual environment (the `(.venv)` prefix).
5. Run one of the commands below.

Windows PowerShell:

```powershell
cd path	oike_mujoco_sim
.\.venv\Scripts\Activate.ps1
python run_sim.py --config configs/tutorial.yaml --model new_bike --mode speed-schedule
python run_sim.py --config configs/tutorial.yaml --model new_bike --mode remote
python run_sim.py --config configs/tutorial.yaml --model new_bike --headless --duration 6
```

macOS or Linux:

```bash
cd path/to/bike_mujoco_sim
source .venv/bin/activate
python run_sim.py --config configs/tutorial.yaml --model new_bike --mode speed-schedule
python run_sim.py --config configs/tutorial.yaml --model new_bike --mode remote
python run_sim.py --config configs/tutorial.yaml --model new_bike --headless --duration 6
```

What the flags mean:

- `--config` — which settings file
- `--model new_bike` — which 3D bike
- `--mode speed-schedule` — just cruise (no stick)
- `--mode remote` — 3D window plus joystick
- `--headless` — no window, print a short report (good for tests)


## Optional check still inside Jupyter

This is **not** a substitute for the terminal commands. It only asks: "if I
load the YAML file right now, does the bike stay up for 6 seconds?" If your
YAML is still the weak 5 / 0 / 6 set, the cell uses the known-good
`new_bike.yaml` so **Run All** does not fail while you are still editing.


````python
tutorial_yaml = os.path.join(project_root, "configs", "tutorial.yaml")
new_bike_yaml = os.path.join(project_root, "configs", "new_bike.yaml")
cfg = load_config(tutorial_yaml)
# If the student has not saved tuned gains yet, fall back so Run All still works.
if cfg.pid_kd_roll < 8.0:
    print("tutorial.yaml still weak; verifying with new_bike.yaml for this check")
    cfg = load_config(new_bike_yaml)
else:
    print("verifying tutorial.yaml")
sim = make_sim(model_spec="new_bike", mode="speed-schedule", cfg=replace(cfg, duration=6.0))
history = run_headless(sim, duration=6.0)
summarize(history, model_path=sim.model_path, mode="deployment-check")
max_roll = float(np.max(np.abs(np.degrees(history["roll"]))))
print(f"max |roll| = {max_roll:.2f} deg")
if max_roll >= 20.0:
    raise RuntimeError("saved gains did not keep the bike upright")
````
