# Teleoperation and disturbance

**Teleoperation** means a human drives the bike: keyboard, on-screen stick,
or a gamepad. The balance software stays on. You are not replacing the PID.
You are adding extra commands on top of it.

Run the next cell first (setup). Then read the stick map, optionally open
the 3D window, then look at the comparison plots.

## Why this is interesting

The PID's job is "keep roll near 0°." When you turn left, you are asking
the handlebars to move *and* the bike to change heading. That extra request
is a **disturbance**: something that would knock the bike over if the PID
did not fight it.

If the PID is well tuned, you can turn and change speed while roll stays
small. That is **disturbance rejection** — a fancy name for "keep balancing
while other things happen."

## How the stick is mapped

Imagine a joystick that **springs back to the middle** when you let go.
Middle means "do nothing extra."

**Up / down (stick Y)** — speed. Middle means "cruise at the usual speed"
(about 2.2 m/s), which is the same idea as the speed schedule, so the bike
can stay up. Full forward asks for a higher speed. Full back asks for a
lower (even reverse) speed. A tiny motion is ignored (a **deadzone**) so
noise does not twitch the speed.

**Left / right (stick X)** — turn. While you **hold** left, the software
adds a turn-rate on top of the balance PID. When you **release**, it
remembers the current heading (which way the nose points) and tries to
keep going that way.

**0** — stop (speed 0). The bike will fall if you leave it stopped; that is
expected. Move up/down off center to leave stop.

**Space or Reset** — put the bike back upright after a fall.

The roll PID **never turns off**. Your stick is an extra input it must live
with.


```python
# Prefer a tuned panel. Fall back to new_bike.yaml if tutorial.yaml is still weak.
tutorial_yaml = os.path.join(project_root, "configs", "tutorial.yaml")
new_bike_yaml = os.path.join(project_root, "configs", "new_bike.yaml")
cfg = load_config(tutorial_yaml)
if cfg.pid_kd_roll < 8.0 or cfg.pid_kp_roll < 6.0:
    cfg = load_config(new_bike_yaml)
    print("loaded tuned gains from", os.path.relpath(new_bike_yaml, project_root))
else:
    print("loaded", os.path.relpath(tutorial_yaml, project_root))
print(
    "outer gains kp={:.3f} ki={:.3f} kd={:.3f} kp_steer={:.3f}".format(
        cfg.pid_kp_roll, cfg.pid_ki_roll, cfg.pid_kd_roll, cfg.pid_kp_steer
    )
)
print("center stick speed goal", speed_goal_from_stick(0.0, cfg))
print("full forward", speed_goal_from_stick(1.0, cfg))
print("full back", speed_goal_from_stick(-1.0, cfg))
```

## Same idea as two plots (no window)

We run two 4-second experiments with the **same** tuned gains:

- stick in the middle (cruise, no extra turn)
- stick held fully left (`stick_x = 1`)

Each plot is roll versus time. Both should stay far from ±45°. The turning
plot may bounce a bit more — that extra bounce is the disturbance. If the
turning plot still stays within about 30°, the PID is doing its job: you
asked for a turn, and it still kept the bike from falling.


```python
def run_remote(stick_x, seconds=4.0):
    sim = make_sim(model_spec="new_bike", mode="remote", cfg=replace(cfg, duration=seconds))
    sim.goals.stick_x = float(stick_x)
    history = run_headless(sim, duration=seconds)
    return sim, history

straight_sim, straight = run_remote(0.0)
turn_sim, turning = run_remote(1.0)
summarize(straight, model_path=straight_sim.model_path, mode="remote center")
summarize(turning, model_path=turn_sim.model_path, mode="remote stick_x=1")

fig, axes = plt.subplots(1, 2, figsize=(10, 3.5), sharey=True, constrained_layout=True)
for ax, hist, title in (
    (axes[0], straight, "Stick centered (cruise)"),
    (axes[1], turning, "Stick held left (disturbance)"),
):
    ax.plot(hist["time"], np.degrees(hist["roll"]))
    ax.axhline(0.0, color="0.7", linewidth=0.8)
    ax.set_title(title)
    ax.set_xlabel("time (s)")
    ax.set_ylabel("roll (deg)")
plt.show()

turn_roll = float(np.max(np.abs(np.degrees(turning["roll"]))))
print(f"max |roll| while turning = {turn_roll:.1f} deg")
if turn_roll > 30.0:
    raise RuntimeError("turn leaned too far; check that tuned gains are loaded")
```

## Cruise with no extra steering

This run uses the speed schedule again: only the speed ramp, no stick turn.
Roll should stay small while speed climbs to cruise. Compare it mentally to
the turning plot: less "wiggle" when you are not commanding a turn.


```python
cruise = make_sim(model_spec="new_bike", mode="speed-schedule", cfg=replace(cfg, duration=6.0))
cruise_hist = run_headless(cruise, duration=6.0)
summarize(cruise_hist, model_path=cruise.model_path, mode="speed-schedule")
fig, ax = plt.subplots(figsize=(8, 3.2), constrained_layout=True)
ax.plot(cruise_hist["time"], np.degrees(cruise_hist["roll"]), label="roll")
ax.set_ylabel("roll (deg)")
ax.set_xlabel("time (s)")
ax2 = ax.twinx()
ax2.plot(cruise_hist["time"], cruise_hist["speed"], color="C1", label="speed")
ax2.set_ylabel("speed (m/s)")
ax.set_title("Constant-speed schedule with tuned balance")
plt.show()
```

## What this showed

A turn or a speed change is extra work for a loop whose main job is "stay
upright." A good PID keeps roll small while those extras happen. That is
the same idea as riding a real bike: you can steer and still not fall,
because balance and steering happen together.
